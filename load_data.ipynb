```python
import json
import pandas as pd

# Paths
corpus_path = "dataset/corpus.jsonl"
queries_path = "dataset/queries.jsonl"
qrels_path = "dataset/qrels/test.tsv"

# ---- Helper function to preview JSONL ----
def preview_jsonl(file_path, n=5):
    print(f"\n--- Previewing {file_path} ---")
    with open(file_path, "r", encoding="utf-8") as f:
        for i, line in enumerate(f):
            if i >= n:
                break
            print(json.loads(line))

# ---- Helper function to preview TSV ----
def preview_tsv(file_path, n=5):
    print(f"\n--- Previewing {file_path} ---")
    df = pd.read_csv(file_path, sep="\t")
    print(df.head(n))



```

```python
preview_jsonl(corpus_path)
```

```python
len(preview_jsonl(corpus_path))
```

```python
with open(corpus_path, "r", encoding="utf-8") as f:
    for i, line in enumerate(f):
        if i >= 5:
            break

        record = json.loads(line)

        print(f"Record {i+1}")
        print("ID:", record.get("_id"))
        print("Title:", record.get("title"))
        print("Text (first 200 chars):", record.get("text", "")[:200], "...")
        print("URL:", record.get("metadata", {}).get("url"))
        print("PubMed ID:", record.get("metadata", {}).get("pubmed_id"))
        print("-" * 60)
```

```python
import json
import random

input_path = "dataset/corpus.jsonl"
output_path = "dataset/corpus_20_percent.jsonl"

sample_ratio = 0.2

with open(input_path, "r", encoding="utf-8") as f_in, \
     open(output_path, "w", encoding="utf-8") as f_out:

    for line in f_in:
        if random.random() < sample_ratio:
            f_out.write(line)

print("Sampling complete → saved to corpus_20_percent.jsonl")
```
